```
# 1) Descargar el proyecto e instalar dependencias (~2 min)
import os
if not os.path.exists('/content/sigma-vial'):
    !git clone -q https://github.com/lexxvc/sigma-vial.git /content/sigma-vial
%cd /content/sigma-vial
!pip install -q -r requirements.txt
```

```
# 2) Clave de Gemini (opcional: sin ella la consola usa el guion fijo)
try:
    from google.colab import userdata
    os.environ['GEMINI_API_KEY'] = userdata.get('GEMINI_API_KEY')
    os.environ['SIGMA_LLM_DEFECTO'] = 'gemini'
    print('GEMINI_API_KEY cargada')
except Exception:
    print('Sin GEMINI_API_KEY: se usará el motor de guion fijo')
```

```
# 3) Arrancar la API + consola y abrirla
import subprocess, time
subprocess.Popen(['python', '-m', 'uvicorn', 'api.main:app', '--port', '8000'], stdout=open('api.log', 'w'), stderr=subprocess.STDOUT)
time.sleep(8)
from google.colab import output
output.serve_kernel_port_as_window(8000, anchor_text='👉 Abrir la consola SIGMA-Vial')
```

Deja esta pestaña abierta mientras usas la consola. Si algo falla, revisa `api.log`.
